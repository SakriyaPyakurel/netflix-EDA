# Data inspection and cleaning 

In this notebook we will inspect and throughly apply cleaning techniques for the dataset so that we can properly utilize EDA(Exploratory Data Analysis)  

```python
# load the dataset 
import pandas as pd 
df = pd.read_csv('netflix_titles.csv') 
```

```python
# View the first few rows of the dataset 
df.head() 
```

Now let's get on dealing with missing values in the dataset and handle them effectively

```python
# checking for missing values 
df.isnull().sum() 
```

<b>Interpretations:</b><br>
1. From the above results we can signify the amount of missing values in each colummn of the dataset now lets get on with handling them<br>
2. As we can notice all the columns with missing values are non-numeric therefore we should drop the rows with missing values rather than replacing them<br>
3. missing countries  and director values will be replaced with Unknown and missing cast,rating and date_added values will be replaced with Unavailable

```python
# Handling missing values 
df['country'].fillna('Unknown', inplace=True) 
df['director'].fillna('Unknown', inplace=True) 
df['cast'].fillna('Unavailable', inplace=True)
df['rating'].fillna('Unavailable', inplace=True) 
df['date_added'].fillna('Unavailable', inplace=True)
```

```python
# viewing the dataset upto 10 rows after handling missing values 
df.head(10)
```

Therefore all the missing values in the dataset are handled

```python
# Converting the date_added column to datetime format 
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce') 
df['release_year'] = pd.to_datetime(df['release_year'], errors='coerce').dt.year
```

```python
# viewing the dataset after converting date_added and release_year columns 
df.head(10)
```

# Basic Visualizations

We will use matplotlib and seaborn for basic visualizations of data to get better interpretations of the data and gather insights about it.

```python
# importing the necessary libraries for visualization 
import seaborn as sb 
import matplotlib.pyplot as plt 
```

```python
# Counting the number of titles by type (movie or TV show) 
print("Number of Movies:", df[df['type'] == 'Movie'].shape[0]) 
print("Number of TV Shows:", df[df['type'] == 'TV Show'].shape[0]) 
```

```python
# Movies vs TV show - graph 
sb.countplot(x='type',data=df)
```

<b>Interpretation:</b> Without looking at the numerical results we can directly look onto above graph and say that the number of movies is more than double of the number of TV Shows.  

Exploring the overview of number of content added per year through visualization 

```python
# Content added per year - graphical overview 
df['release_year'].value_counts().sort_index().plot(kind='bar', figsize=(12,10), color='skyblue') 
plt.title('Number of Titles Added per Year')
plt.xlabel('Year') 
plt.ylabel('Number of Titles') 
```

<b>Interpretations:</b><br>
1. The number of content(TV shows and movies) added each year are increasing from 1925 onwards.<br>
2. The start of 21st century(2000's) has been seen marking the start of increased content releases.<br>
3. Year 2014 marked the remarkable increase of content release may be attributed to the global expansion of platforms like Netflix, Amazon Prime, etc.<br> 
4. The production halts and uncertainty during the pandemic likely contributed to the decline in new content additions post-2019.

Exploring top 10 countries by content 

```python
df['country'].value_counts().head(10)
```

```python
# Visualizing the top 10 countries by content 
top_countries = df['country'].value_counts().head(10)
top_countries.plot(kind='bar', figsize=(12, 6), color='lightgreen')
plt.title('Top 10 Countries by Content') 
plt.xlabel('Country') 
plt.ylabel('Number of Titles')
plt.xticks(rotation=45) 
plt
```

<b>Interpretations:</b><br> 
1. United states is seen being the highest creator of contents whereas egypt stands low in the list. <br>
2. Supremacy of united states in content creation and release can be marked up with its availability of technology and democratic-capitalistic nature.<br>
3. India ranked as second in the list is most likely due to its super high population and demand of entertaining contents.<br>
4. Countries like the UK, Japan, and South Korea also produce a significant amount of content, but their lower representation may reflect licensing decisions or content localization preferences on the platform.

Exploring content ratings distribution 

```python
df['rating'].value_counts()
```

```python
# Visualizing the distribution of ratings 
sb.countplot(y='rating', data=df, order=df['rating'].value_counts().index, palette='muted')
```

<b>Interpretations:</b><br>
1. TV-MA(Mature Audience) content has highest number of releases indicating the continous upsurged demand for mature content.<br>
2. TV-14 is the second most highest number of released content type with slightly less maturity than TV-MA but still designed for 14+ Teens.<br>
3. Ratings like TV-Y, TV-G, G, PG, and TV-Y7-FV are far lower in comparison indicating less dominance of kids and family content in the platform.<br>
4. "74 min", "84 min", "66 min" are most likely data entry issues where duration values ended up in the rating column.<br>
5. Ratings like NR(Non-Rated), UR(Unrated) and NC-17 are very few which is expected due to their niche or restricted nature.
