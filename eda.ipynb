# Data inspection and cleaning 

In this notebook we will inspect and throughly apply cleaning techniques for the dataset so that we can properly utilize EDA(Exploratory Data Analysis)  

```python
# load the dataset 
import pandas as pd 
df = pd.read_csv('netflix_titles.csv') 
```

```python
# View the first few rows of the dataset 
df.head() 
```

Now let's get on dealing with missing values in the dataset and handle them effectively

```python
# checking for missing values 
df.isnull().sum() 
```

<b>Interpretations:</b><br>
1. From the above results we can signify the amount of missing values in each colummn of the dataset now lets get on with handling them<br>
2. As we can notice all the columns with missing values are non-numeric therefore we should drop the rows with missing values rather than replacing them<br>
3. missing countries  and director values will be replaced with Unknown and missing cast,rating and date_added values will be replaced with Unavailable

```python
# Handling missing values 
df['country'].fillna('Unknown', inplace=True) 
df['director'].fillna('Unknown', inplace=True) 
df['cast'].fillna('Unavailable', inplace=True)
df['rating'].fillna('Unavailable', inplace=True) 
df['date_added'].fillna('Unavailable', inplace=True)
```

```python
# viewing the dataset upto 10 rows after handling missing values 
df.head(10)
```

Therefore all the missing values in the dataset are handled

```python
# Converting the date_added column to datetime format 
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce') 
df['release_year'] = pd.to_datetime(df['release_year'], errors='coerce').dt.year
```

```python
# viewing the dataset after converting date_added and release_year columns 
df.head(10)
```

Distribution of genres between Movies and TV Shows

```python
# Extracting top genres after importing Counter from collections library 
from collections import Counter 
# Flatten genres 
flat_genres = [genre.strip() for sublist in df['listed_in'].dropna().apply(lambda x: x.split(',')) for genre in sublist]
genre_counts = pd.Series(Counter(flat_genres)).sort_values(ascending=False).head() 
genre_counts
```

```python
# Add top genre to each row 
df['main_genre'] = df['listed_in'].apply(lambda x: next((g for g in genre_counts.index if g in x),'Other')) 
df.head()
```

```python
sb.countplot(data=df,y='main_genre',hue='type') 
plt.title('Content Type by Top Genres')
```

<b>Interpretations:</b><br>

1. Compared how genres are distributed between Movies and TV Shows.<br>
2. Movies are leading in all genre counts.<br>
3. Excluding Internationals, other genres in both movies are tv shows are leading indicating massive deviation of consumers from regualar to new type of content genres.<br>
4. International movies are most in number within the platform.<br>

C. Multivariate Analysis: Useful for gathering layered insights with three or more variables.

i. 'Content Type' & 'Genre' & 'Country'

```python
# Top 5 countries and Genres 
top_countries = df['country'].value_counts().head(5).index 
top_genres = df['main_genre'].value_counts().head(5).index 
filtered_df = df[df['country'].isin(top_countries) & df['main_genre'].isin(top_genres)] 
```

```python
# Plotting 
sb.catplot(data=filtered_df,kind='count',x='country',hue='main_genre',col='type',height=5,aspect=1.5) 
plt.subplots_adjust(top=0.8) 
plt.suptitle('Genre distribution by Country and Type')
```

<b>Interpretations:</b><br>
TV shows:<br>
i. United states is seen being highest in Other type of TV shows while being lowest in international TV shows indicating the massive fleeting interest of audience there.<br>
ii. Unknown type of TV show content are mostly identical in distribution.<br>
iii. UK(United Kingdom) have an stopping curve shaped data distribution from left to right with international shows are most made while Dramas are least made.<br>
iv. India got least made other type of tv shows which shows static interest of the audience there.<br>
v. Japan has lowest number of TV shows in comedy genre implying the interest of people in more serious type of tv show content.<br>

Movies:<br>

i. India is seen as the highest creator and watcher of international movies indicating their desire to connect more with the outside world.<br>
ii. Unknown international movies are seen being highest indicating secrecy of crews while creating international movies.<br>
iii. Other type of movies are most high ranked in United states indicating the fleeting and explorative interest of american audience just like in that of TV shows.<br>
iv. United Kingdom have almost equal frequency of comedy and dramatic movies indicating the possible chill working style and work-life balance of the audience there.<br>
v. Japan lacks in availability of comedy and drama type of content again indicating the seriousness of audience when it comes to digital content while international movies are top in number almost similar in distribution to that of United Kingdom.

```python
# Grouping by rating and getting average duration
rating_duration = movies.groupby('rating')['duration_cleaned'].mean().sort_values(ascending=False) 
rating_duration.plot(kind='bar',color='brown') 
plt.ylabel('Avg. Duration')
plt.title('Average Movie Duration by Rating')
```

<b>Interpretations:</b><br>

i. The most dominant rating i.e. TV-MA tend to moderately long (about '1 hour 30 mins' or '90 mins') indicating the liking of matured audience for content with average durations.<br>
ii. Second most dominant rating i.e. R(Rated) and UR(Unrated) tend to crossing the 100 mins mark indicating the possibility of suspense and long thrills.<br>
iii. NC-17 content considered suitable only for adult audiences, aged 18 and over are seen having more durations possibly due to their mature themes and visuals attracting the minds of most audiences.<br>
iv. "66 min", "74 min", "84 min" appear as ratings but clearly are mistakenly parsed durations likely came from misclassified data during preprocessing

```python
df_country_year = df[df['country'].isin(top_countries)].copy()
df_country_year['year_added'] = pd.to_numeric(df_country_year['year_added'], errors='coerce')
agg = df_country_year.groupby(['year_added', 'country']).size().reset_index(name='count')
sb.lineplot(data=agg, x='year_added', y='count', hue='country')
plt.title('Content Addition Trends over Time by Country')
plt.xlabel('Year Added')
plt.ylabel('Number of Titles')
```

iv. Unkown country's contents have rised up significantly in number from early 2016 to a substantial growth from 2020 onwards indicating the time of covid lockdown and free time people had to watch any type of available content.<br>
