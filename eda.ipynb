# Data inspection and cleaning 

In this notebook we will inspect and throughly apply cleaning techniques for the dataset so that we can properly utilize EDA(Exploratory Data Analysis)  

```python
# load the dataset 
import pandas as pd 
df = pd.read_csv('netflix_titles.csv') 
```

```python
# View the first few rows of the dataset 
df.head() 
```

Now let's get on dealing with missing values in the dataset and handle them effectively

```python
# checking for missing values 
df.isnull().sum() 
```

<b>Interpretations:</b><br>
1. From the above results we can signify the amount of missing values in each colummn of the dataset now lets get on with handling them<br>
2. As we can notice all the columns with missing values are non-numeric therefore we should drop the rows with missing values rather than replacing them<br>
3. missing countries  and director values will be replaced with Unknown and missing cast,rating and date_added values will be replaced with Unavailable

```python
# Handling missing values 
df['country'].fillna('Unknown', inplace=True) 
df['director'].fillna('Unknown', inplace=True) 
df['cast'].fillna('Unavailable', inplace=True)
df['rating'].fillna('Unavailable', inplace=True) 
df['date_added'].fillna('Unavailable', inplace=True)
```

```python
# viewing the dataset upto 10 rows after handling missing values 
df.head(10)
```

Therefore all the missing values in the dataset are handled

```python
# Converting the date_added column to datetime format 
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce') 
df['release_year'] = pd.to_datetime(df['release_year'], errors='coerce').dt.year
```

```python
# viewing the dataset after converting date_added and release_year columns 
df.head(10)
```

# Getting Started with EDA(Explaratory Data Analysis) 

1. Univariate Analysis: Exploring distributions of individual variables

We will use matplotlib and seaborn for basic visualizations of data to get better interpretations of the data and gather insights about it.

```python
# importing the necessary libraries for visualization 
import seaborn as sb 
import matplotlib.pyplot as plt 
```

```python
# Counting the number of titles by type (movie or TV show) 
print("Number of Movies:", df[df['type'] == 'Movie'].shape[0]) 
print("Number of TV Shows:", df[df['type'] == 'TV Show'].shape[0]) 
```

```python
# Movies vs TV show - graph 
sb.countplot(x='type',data=df)
```

<b>Interpretation:</b> Without looking at the numerical results we can directly look onto above graph and say that the number of movies is more than double of the number of TV Shows.  

Exploring the overview of number of content added per year through visualization 

```python
# Content added per year - graphical overview
df['release_year'] = pd.to_numeric(df['release_year'],errors='coerce')
year_counts = df['release_year'].dropna().astype(int).value_counts().sort_index()
plt.figure(figsize=(14, 6))
plt.bar(year_counts.index.astype(str), year_counts.values, color='skyblue')
plt.title('Number of Titles Added per Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.xticks(rotation=90)
plt.tight_layout()
plt
```

<b>Interpretation</b> All the movies and tv_shows are released in 1970.

4. Some movies having overall duration up to more than 200 minutes(3 hour 20 mins) are also seen existing in the platforms which maybe most likely documentary type content.

Distribution of genres between Movies and TV Shows

```python
# Extracting top genres after importing Counter from collections library 
from collections import Counter 
# Flatten genres 
flat_genres = [genre.strip() for sublist in df['listed_in'].dropna().apply(lambda x: x.split(',')) for genre in sublist]
genre_counts = pd.Series(Counter(flat_genres)).sort_values(ascending=False).head() 
genre_counts
```

```python
# Add top genre to each row 
df['main_genre'] = df['listed_in'].apply(lambda x: next((g for g in genre_counts.index if g in x),'Other')) 
df.head()
```

```python
sb.countplot(data=df,y='main_genre',hue='type') 
plt.title('Content Type by Top Genres')
```

<b>Interpretations:</b><br>

1. Compared how genres are distributed between Movies and TV Shows.<br>
2. Movies are leading in all genre counts.<br>
3. Excluding Internationals, other genres in both movies are tv shows are leading indicating massive deviation of consumers from regualar to new type of content genres.<br>
4. International movies are most in number within the platform.<br>

ii. Categorical and Numerical

1. Rating and Average movie duration

```python
movies['duration_cleaned'] = movies['duration'].dropna().str.replace(' min', '').astype(float)
# Box plotting 
sb.boxplot(data=movies,x='rating',y='duration_cleaned') 
plt.title('Movie Duration by Rating') 
plt.xticks(rotation=45)
```

<b>Interpretations:</b><br>
1. The rectangular boxes represent the spread of all ratings.<br>
2. The circle structure represent outliers or abnormalities for each rating which are to be erased for optimality.<br>
3. TV-MA(Matured Audience) rating has maximum outliers on both sides whereas NR and TV-G have minimum outliers.<br>
4. There are no rectangles for '74 min','84 min' and '66 min' indicating zero spread in the data.<br>
5. Ratings- Unavailable,TV-Y7-FV and UR(Unrated) are prefect with no outliers.
